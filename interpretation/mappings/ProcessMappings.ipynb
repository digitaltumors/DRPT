# Process Mappings

A helper notebook used to do any data processing for the mappings files. In particular, this notebook will process the `NEST_Annotations.txt` file for nest-to-name and system-to-gene files

```python
import pandas as pd
import numpy as np
import pickle
```

```python
annotations = pd.read_csv("NEST_Annotations.txt", sep = '\t')
annotations.head()
```

## Nest-to-Old Name and System-to-Genes

```python
nest_to_name = {}
system_genes = {}
for ind in annotations.index:
    nest = str(annotations["ID"][ind]).upper()
    name = str(annotations["Annotation"][ind])
    genes = str(annotations["All_Genes"][ind]).split(",")

    nest_to_name[nest] = name
    system_genes[name] = genes
```

```python
len(system_genes)
```

### Re-save files

```python
with open("nest_to_old_name.pkl", "wb") as handle:
    pickle.dump(nest_to_name, handle)

with open("system_to_genes.pkl", "wb") as handle:
    pickle.dump(system_genes, handle)
```
